```python
## Perform Logistic Regression 
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

X_train = df.drop('Activity', axis=1)
y_train = df['Activity']

X_test = testdf.drop('Activity', axis=1)
y_test = testdf['Activity']

le = LabelEncoder()
y_train_encoded = le.fit_transform(y_train)
y_test_encoded = le.transform(y_test)
```

```python
log_reg = LogisticRegression(max_iter=1000)
log_reg.fit(X_train, y_train_encoded)
```

```python
y_pred = log_reg.predict(X_test)
score = accuracy_score(y_test_encoded, y_pred)
print(score)
```

```python
def get_duplicate_columns(df):

    duplicate_columns = {}
    seen_columns = {}

    for column in df.columns:

        current_column = df[column]

        try:
            current_column_hash = current_column.values.tobytes()
        except AttributeError:
            current_column_hash = current_column.to_string().encode()

        if current_column_hash in seen_columns:
            first_column = seen_columns[current_column_hash]
            if first_column in duplicate_columns:
                duplicate_columns[first_column].append(column)
            else:
                duplicate_columns[first_column] = [column]
        else:
            seen_columns[current_column_hash] = column
    return duplicate_columns
```

```python
duplicate_column = get_duplicate_columns(X_train)
duplicate_column
```

```python
for first_col, dup_cols in duplicate_column.items():
    for col in dup_cols:
        X_train = X_train.drop(columns=col, errors='ignore')
        X_test = X_test.drop(columns=col, errors='ignore')
```

```python
X_test.columns.value_counts()
```

```python
from sklearn.feature_selection import VarianceThreshold
sel = VarianceThreshold(threshold=0.05)
```

```python

sel.fit(X_train)
```

```python

sum(sel.get_support())
```

```python
# Preserve selected column names from the DataFrame
columns = X_train.columns[sel.get_support()]
columns
```

```python
X_train_selected = sel.transform(X_train)
X_test_selected = sel.transform(X_test)

X_train_selected = pd.DataFrame(
    X_train_selected,
    columns=columns,
    index=X_train.index
)

X_test_selected = pd.DataFrame(
    X_test_selected,
    columns=columns,
    index=X_test.index
)
```

```python
X_train = X_train_selected
X_test = X_test_selected
```

```python
columns = corr_matrix.columns

columns_to_drop = []

for i in range(len(columns)):
    for j in range(i + 1, len(columns)):
        if abs(corr_matrix.loc[columns[i] ,columns[j]]) > 0.95:
            columns_to_drop.append(columns[j])

len(columns_to_drop)
```

```python
columns_to_drop = set(columns_to_drop)
len(columns_to_drop)
```

```python
X_train.drop(columns= columns_to_drop, axis = 1, inplace = True)
X_test.drop(columns= columns_to_drop, axis = 1, inplace = True)
```
